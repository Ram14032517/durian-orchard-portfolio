```python
from pathlib import Path
import pandas as pd
root=Path.cwd()
if not (root/'research_data').exists():
    root=root.parents[1]
p=root/'research_data/five_province_history'
a=pd.read_csv(p/'production_weather_annual.csv',dtype={'province_code':str})
d=pd.read_csv(p/'weather_daily.csv',dtype={'province_code':str})
s=pd.read_csv(p/'soil_properties_by_district.csv',dtype={'province_code':str})
assert not d.duplicated(['province_code','date']).any()
assert a.province_code.nunique()==5
print('อากาศรายวัน:',len(d),'ระเบียน; ผลผลิต:',len(a),'จังหวัด–ปี')
```
